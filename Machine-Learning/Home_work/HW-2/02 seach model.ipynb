```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import RobustScaler, OneHotEncoder, PowerTransformer, PolynomialFeatures
from sklearn.impute import SimpleImputer
from sklearn.svm import SVR
from sklearn.metrics import mean_squared_log_error, make_scorer
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
from sklearn.compose import TransformedTargetRegressor
```

### Область работы 2 (поиск  модели .... )

```python
# Данный блок(и) НЕ выполняются преподавателем в области 2
# блок(и) предназначены для поиска лучшей модели 
# должен быть понятен и очевиден отбор параметров модели
# 
# Запрещается размещать данные блоки за пределами обасти 2
# Все блоки данной области должны быть выполнены
#
# ЗАПРЕЩАЕТСЯ ИСПОЛЬЗОВАТЬ ТЕСТОВЫЙ НАБОР
#
# Путь к тренировочному набору
# 

path_train = 'train.csv' # содержит только имя файла, без имен папок !!!
```

```python
df = pd.read_csv(path_train)
df
```

```python
drop_cols = [
    'fancy_color_dominant_color', 'fancy_color_secondary_color', 
    'fancy_color_overtone', 'fancy_color_intensity',
    'culet_condition', 'fluor_color'
]
df = df.drop(columns=drop_cols, errors='ignore')
df
```

```python
# проверка дубликатов
print("Количество дубликатов:", df.duplicated().sum())
df = df.drop_duplicates()
```

```python
num_features = [
    'size', 'depth_percent', 'table_percent', 
    'meas_length', 'meas_width', 'meas_depth'
]
cat_features = [
    'color', 'clarity', 'cut', 'symmetry', 'polish', 
    'girdle_min', 'girdle_max', 'fluor_intensity', 'eye_clean'
]
```

```python
X = df.drop(columns=['total_sales_price'])
y = df['total_sales_price']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.33
)
```

```python
num_transformer = Pipeline(steps=[
    ('imputer', IterativeImputer(max_iter=20)),
    ('poly', PolynomialFeatures(2, include_bias=False)),
    ('scaler', RobustScaler())
])

cat_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(sparse_output=False, handle_unknown='ignore'))
])
```

```python
CT = ColumnTransformer([
    ('num', num_transformer, num_features),
    ('cat', cat_transformer, cat_features)
]).set_output(transform='pandas')
```

```python
pipeline = Pipeline(steps=[
    ('preproc', CT),
    ('estimator', SVR())
])

model_target = TransformedTargetRegressor(
    regressor=pipeline,
    transformer=PowerTransformer(method='yeo-johnson')
)

def msle_scorer(y_true, y_pred):
    return -mean_squared_log_error(y_true, np.maximum(y_pred, 0))

score_msle = make_scorer(mean_squared_log_error, greater_is_better=False)

# Обучение
model_target.fit(X_train, y_train)
y_pred = model_target.predict(X_test)

msle = mean_squared_log_error(y_test, np.maximum(y_pred, 0))
print(f"Mean Squared Logarithmic Error (MSLE): {msle:.6f}")
```
