## Каталог для скачанных документов

```python

SAVE_DIR = Path("ranepa_documents")
SAVE_DIR.mkdir(exist_ok=True)


def download_all():
    """
    Скачиваем все ссылки из document_links в папку SAVE_DIR.
    Если файл уже существует — пропускаем. ZIP-файлы дополнительно распаковываем.
    """
    for url in document_links:
        fname = unquote(Path(url).name)
        path = SAVE_DIR / fname

        # Пропускаем, если скачан ранее
        if path.exists():
            print(f"⏩ Уже скачано: {fname}")
            continue

        try:
            resp = requests.get(url, timeout=60)
            resp.raise_for_status()
            path.write_bytes(resp.content)
            print(f"✅ Скачан: {fname}")

            # Если файл — ZIP, распакуем в папку с его именем (без расширения)
            if fname.lower().endswith(".zip"):
                extract_to = SAVE_DIR / path.stem
                extract_to.mkdir(exist_ok=True)
                with ZipFile(path, "r") as zf:
                    zf.extractall(extract_to)
                print(f"📦 Распакован: {fname} → {extract_to}")

        except Exception as e:
            print(f"⚠️ Не удалось скачать {fname}: {e}")


# =============================================================================
```

##  Конвертация PDF / DOCX / HTML → plain text

```python

from PyPDF2 import PdfReader
from docx import Document as DocxDocument


def pdf_to_text(fp: Path) -> str:
    """
    Извлекаем текст со всех страниц PDF с помощью PyPDF2.
    Если страница дала None → заменяем на пустую строку.
    """
    reader = PdfReader(str(fp))
    text_pages = [page.extract_text() or "" for page in reader.pages]
    return "\n".join(text_pages)


def docx_to_text(fp: Path) -> str:
    """
    Извлекаем текст из DOCX-параграфов с помощью python-docx.
    """
    doc = DocxDocument(str(fp))
    return "\n".join(p.text for p in doc.paragraphs)


def plain_text(fp: Path) -> str:
    """
    По суффиксу файла (.pdf/.docx/.txt/.html/.htm) возвращаем разными методами чистый текст.
    """
    suffix = fp.suffix.lower()
    if suffix == ".pdf":
        return pdf_to_text(fp)
    if suffix == ".docx":
        return docx_to_text(fp)
    if suffix in {".txt", ".html", ".htm"}:
        return fp.read_text(encoding="utf-8", errors="ignore")
    return ""


# =============================================================================
```

## Смысловое разбиение (TF‑IDF + кластеризация)

```python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import AgglomerativeClustering


def split_sentences(text: str):
    """
    Простейшее разбиение на предложения с помощью regex:
      (?<=[.!?])\s+
    → разделяем после точки/вопросительного/восклицательного знака и пробелов.
    """
    parts = re.split(r"(?<=[.!?])\s+", text)
    return [p.strip() for p in parts if p.strip()]


def semantic_chunks(sentences, max_sent_per_chunk=5):
    """
    Алгоритм «сенс-чанков»:

    1) Векторизуем каждое предложение TfidfVectorizer(max_features=5000).
    2) Находим «нулевые» векторы строк (они оказываются pure-zero,
       если в предложении только стоп-слова или токены не вошли в словарь).
    3) Если все — «нулевые», возвращаем один чанк из всего текста.
    4) Иначе:
            - Кластеризуем только ненулевые векторы (с AgglomerativeClustering
            с affinity="cosine", linkage="average").
            - Составляем список labels: ненулевым предложениям присваиваем
            метки от кластеризации, «нулевым» присваиваем метку −1.
            - Проходим по исходному порядку предложений и склеиваем подряд
            идущие предложения, у которых одинаковая метка.
    """
    if not sentences:
        return []

    # 1) TF‑IDF
    vectorizer = TfidfVectorizer(max_features=5000)
    mat = vectorizer.fit_transform(sentences)
    dense = mat.toarray()  # shape = (n_sentences, n_features)

    # 2) Определяем, какие строки — полностью нулевые
    zero_idx = [i for i, row in enumerate(dense) if not row.any()]
    nonzero_idx = [i for i in range(len(sentences)) if i not in zero_idx]

    # 3) Если все предложения — «нулевые» → возвращаем один чанк (join)
    if len(nonzero_idx) == 0:
        return [" ".join(sentences)]

    # 4.1) Определяем число кластеров: ceil(len(nonzero_idx)/max_sent_per_chunk)
    n_clusters = max(1, math.ceil(len(nonzero_idx) / max_sent_per_chunk))
    # Если кластеров ⇒ >= ненулевых предложений, дробить нет смысла
    if n_clusters >= len(nonzero_idx):
        return [" ".join(sentences)]

    # 4.2) Кластеризуем только ненулевые
    sub_mat = dense[nonzero_idx]
    model = AgglomerativeClustering(
        n_clusters=n_clusters,
        affinity="cosine",
        linkage="average"
    ).fit(sub_mat)
    sub_labels = model.labels_

    # 4.3) Восстанавливаем полный список меток длины len(sentences):
    labels = [-1] * len(sentences)
    for idx, lab in zip(nonzero_idx, sub_labels):
        labels[idx] = lab

    # 4.4) Склеиваем подряд идущие предложения с одинаковой меткой
    chunks = []
    buf = [sentences[0]]
    cur_label = labels[0]
    for sent, lab in zip(sentences[1:], labels[1:]):
        if lab == cur_label:
            buf.append(sent)
        else:
            chunks.append(" ".join(buf))
            buf = [sent]
            cur_label = lab
    if buf:
        chunks.append(" ".join(buf))

    return chunks


# =============================================================================
```

## Главное: собираем DataFrame и сохраняем CSV

```python

def pipeline(limit=None):
    """
    Главная функция:

    1) Проходим по всем файлам в SAVE_DIR (или первые `limit`, если limit задан).
    2) Для каждого файла:
         - извлекаем текст (plain_text),
         - разбиваем на предложения (split_sentences),
         - делаем semantic_chunks(...),
         - для каждого чанкa готовим словарь со столбцами:
           {
             "doc_id":   первые 8 символов MD5 от fp.name,
             "chunk_id": порядковый номер части в этом документе,
             "title":    человекочитаемое название документа (fp.stem, подчёркивания → пробелы),
             "text":     текст чанка (после очистки),
             "url":      исходный URL из filename_to_url[fp.name],
             "doc_type": "Нормативный акт",
             "length":   количество символов в "text"
           }
    3) Собираем ожидание всех таких словарей в list `rows`.
    4) Конвертируем rows → pandas.DataFrame и сохраняем в CSV с BOM (utf-8-sig).
    """
    rows = []
    files = list(SAVE_DIR.rglob("*"))
    if limit:
        files = files[:limit]

    for fp in files:
        if not fp.is_file():
            continue

        text_raw = plain_text(fp)
        if not text_raw.strip():
            # Если пустой результат, пропускаем
            continue

        sentences = split_sentences(text_raw)
        if not sentences:
            continue

        # Получаем sense‑чанки
        chunks = semantic_chunks(sentences, max_sent_per_chunk=5)
        # Генерируем уникальный doc_id
        doc_id = md5(fp.name.encode()).hexdigest()[:8]
        # Читаемая «title» (имя без расширения, подчёркивания → пробелы)
        title = fp.stem.replace("_", " ")
        # Получаем URL из словаря (если файл Zip распакован, url может взять сам ZIP)
        url = filename_to_url.get(fp.name, "")
        # Фиксированный тип
        doc_type = "Нормативный акт"

        for i, chunk in enumerate(chunks):
            # Окончательная очистка «мусорных» номеров страниц:
            # уберём всё, что строкой состоит только из цифр или «Стр. N» / «Страница N»
            lines = chunk.splitlines()
            cleaned_lines = []
            for line in lines:
                s = line.strip()
                if re.fullmatch(r"\d+", s):
                    continue
                if re.fullmatch(r"(Стр\.?|Страница)\s*\d+", s, flags=re.IGNORECASE):
                    continue
                cleaned_lines.append(s)
            text_clean = " ".join(cleaned_lines)

            # Пропускаем, если после чистки мало символов (<100)
            if len(text_clean) < 100:
                continue

            rows.append({
                "doc_id":   doc_id,
                "chunk_id": i,
                "title":    title,
                "text":     text_clean,
                "url":      url,
                "doc_type": doc_type,
                "length":   len(text_clean)
            })

        print(f"   ✅ Обработан: {fp.name} → {len(chunks)} чанков (до фильтрации длиной)")

    # Формируем DataFrame и сохраняем CSV
    df = pd.DataFrame(rows)
    OUT_DIR = Path("data/processed")
    OUT_DIR.mkdir(parents=True, exist_ok=True)
    out_csv = OUT_DIR / "ranepa_semantic_chunks_full.csv"
    df.to_csv(out_csv, index=False, sep=";", quotechar='"', encoding="utf-8-sig")
    print(f"\n💾 Готово: {out_csv}  ({len(df)} строк всего)")


# =============================================================================
#                               ЗАПУСК
# =============================================================================

if __name__ == "__main__":
    pipeline(limit=None)
```
